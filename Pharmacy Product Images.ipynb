```python
from icrawler.builtin import GoogleImageCrawler
import pandas as pd
import os
import time
import re

# Load Excel
df = pd.read_excel("D:\OnyxData\Book1.xlsx")

# Base folder
base_folder = "product_images"
os.makedirs(base_folder, exist_ok=True)

def clean_name(text):
    return re.sub(r'[^a-zA-Z0-9]', '_', text)

for i, row in df.iterrows():
    
    pid = clean_name(str(row["ProductID"]))
    pname = str(row["ProductName"])
    
    print(f"{i+1}/{len(df)} Downloading → {pname}")

    folder_path = f"{base_folder}/{pid}"
    os.makedirs(folder_path, exist_ok=True)

    try:
        crawler = GoogleImageCrawler(
            storage={'root_dir': folder_path},
            feeder_threads=1,
            parser_threads=1,
            downloader_threads=2
        )

        crawler.crawl(
            keyword=pname + " medicine tablet pack",
            max_num=2
        )

        time.sleep(2)   # prevent blocking

    except Exception as e:
        print("Error:", e)
```

```python
from icrawler.builtin import BingImageCrawler
import pandas as pd
import os
import time
import re

# -----------------------------
# SETTINGS
# -----------------------------
EXCEL_FILE = "D:\OnyxData\Book1.xlsx"
PRODUCT_COLUMN = "ProductName"
BASE_FOLDER = "product_images"
IMAGES_PER_PRODUCT = 2
DELAY_SECONDS = 2
# -----------------------------

# Load Excel
df = pd.read_excel(EXCEL_FILE)

print("Columns found:", df.columns.tolist())

if PRODUCT_COLUMN not in df.columns:
    raise ValueError(f"Column '{PRODUCT_COLUMN}' not found in Excel")

# Create base folder
os.makedirs(BASE_FOLDER, exist_ok=True)

# Function to clean folder names
def clean_name(text):
    return re.sub(r'[^a-zA-Z0-9]', '_', str(text))[:80]

# Optional — test first with small sample
# df = df.head(5)

total = len(df)

for i, row in df.iterrows():
    
    pname = str(row[PRODUCT_COLUMN]).strip()
    
    if pname == "" or pname.lower() == "nan":
        print(f"Skipping empty row {i}")
        continue

    folder_name = clean_name(pname)
    folder_path = os.path.join(BASE_FOLDER, folder_name)
    os.makedirs(folder_path, exist_ok=True)

    print(f"\n{i+1}/{total} Downloading → {pname}")

    try:
        crawler = BingImageCrawler(
            storage={'root_dir': folder_path},
            feeder_threads=1,
            parser_threads=1,
            downloader_threads=2
        )

        crawler.crawl(
            keyword=pname + " product packaging",
            max_num=IMAGES_PER_PRODUCT,
            min_size=(200, 200)
        )

        # Check if download worked
        if len(os.listdir(folder_path)) == 0:
            print("⚠️ No images found")

        time.sleep(DELAY_SECONDS)

    except Exception as e:
        print("❌ Error:", e)

print("\n✅ Download process completed")
```

```python
import pandas as pd
import os
import re

EXCEL_FILE = "D:\OnyxData\Book1.xlsx"
BASE_FOLDER = "product_images"

df = pd.read_excel(EXCEL_FILE)

def clean_name(text):
    return re.sub(r'[^a-zA-Z0-9]', '_', str(text))[:80]

image_paths = []

for name in df["ProductName"]:
    
    folder = clean_name(name)
    folder_path = os.path.join(BASE_FOLDER, folder)

    if os.path.exists(folder_path):
        files = os.listdir(folder_path)
        if files:
            first_image = files[0]
            full_path = os.path.abspath(os.path.join(folder_path, first_image))
            image_paths.append(full_path)
        else:
            image_paths.append("")
    else:
        image_paths.append("")

df["ImagePath"] = image_paths

df.to_excel("products_with_images.xlsx", index=False)

print("✅ New Excel file created: products_with_images.xlsx")
```
